Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python
sentences_list = [
    "The city council approved a new plan to build more bicycle lanes across the downtown area, hoping to reduce traffic and improve air quality.",
    "Many people now prefer cycling to work because it saves money on fuel, keeps them fit, and helps them avoid long traffic jams every morning.",
    "Machine learning models can turn sentences into lists of numbers, so that texts with similar meaning end up close to each other in space.",
    "Vector databases store embeddings and let developers search through millions of documents quickly by finding the vectors that are nearest to a given query.",
    "I spent the weekend cooking a big pot of vegetable soup with fresh carrots, potatoes, and herbs from the small garden behind my house.",
    "A good recipe for tomato pasta needs ripe tomatoes, plenty of garlic, olive oil, and a handful of basil added right at the end.",
    "The football team won the championship last night after scoring a late goal, and thousands of fans celebrated in the streets until morning.",
    "Our coach says that regular training, enough sleep, and a balanced diet matter more for young players than expensive equipment or fancy tactics.",
    "Python is a popular programming language for data science because it has simple syntax and many useful libraries for working with numbers and text.",
    "Learning to write clean code takes time, but reading other people's projects and fixing small bugs is a great way to improve quickly.",
]

embeddings = model.encode(sentences_list)
print(embeddings.shape)   # (10, 384)
```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
def cos_similarity_calculation(text, sentences=sentences_list, sent_embeddings=embeddings):
    # embed the new text and compare it with every sentence in the list
    text_emb = model.encode(text)
    scores = util.cos_sim(text_emb, sent_embeddings)[0]

    # show results from most similar to least similar
    results = sorted(zip(sentences, scores.tolist()), key=lambda x: x[1], reverse=True)
    for s, score in results:
        print(f"{score:.3f} | {s[:70]}...")
    return results

# test it
_ = cos_similarity_calculation("How do I make a tasty dinner at home?")
_ = cos_similarity_calculation("Who won the match yesterday?")
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def cosine_vector_vs_batch(vec, batch):
    # cos = (a . b) / (|a| * |b|)
    dots = batch @ vec
    norms = np.linalg.norm(batch, axis=1) * np.linalg.norm(vec)
    return dots / norms

# demo: first sentence against all ten
scores = cosine_vector_vs_batch(embeddings[0], embeddings)
for i, s in enumerate(scores):
    print(f'demo: {i}, {round(float(s), 3)}')

# should match the sentence-transformers result
print(np.allclose(scores, util.cos_sim(embeddings[0], embeddings)[0].numpy(), atol=1e-5))
```

**My Observation:**
- The `paraphrase-MiniLM-L6-v2` model gives every sentence a vector of 384 numbers, no matter how long or short the sentence is. When I encoded my list of 10 sentences, I got an array of shape (10, 384).
- Sentences about the same topic scored higher than unrelated ones. The two cycling sentences got 0.34, but the cycling sentence compared with the football one got only 0.019.
- My `cos_similarity_calculation` function and the numpy function gave the same scores, so both are doing the same cosine calculation.
- When a sentence is compared with itself, the score is always 1.0. Cosine distance is 1 minus the similarity, so a distance of 0 means the two vectors point the same way.
- Cosine similarity only checks the direction of the vectors, not their length. So a long sentence and a short one can still score high if they mean the same thing.
