# 🚀 Day 15 Challenge: Startup in a Box

> *"Every great company started as a crazy idea on a napkin. Your job: build the machine that turns napkins into Demo Day pitches."*

## 📖 The Story

**Moonshot Ventures**, a (fictional) venture studio, receives hundreds of wild product ideas every week. Their analysts can't keep up, so they've hired **you** as their founding AI engineer.

Your mission is to build an **AI venture studio**: a multi-agent system that takes **one wild idea** and turns it into a complete **🚀 Launch Kit**. The kit covers brand, market research, product plan, unit economics, a marketing campaign and an investor pitch, and it has to survive a tough **investor panel** at Demo Day.

You choose the framework. You design the team. You decide how the agents work together. 🧠

## 🎯 Rules of the Game

### ✅ Your system must…
1. **Use one agent framework** from Days 14–15 as its backbone: `smolagents`, `LangGraph`, `LlamaIndex`, `Google ADK` or `CrewAI`.
2. **Have at least 3 agents** with clearly different roles.
3. **Combine at least 2 agentic patterns** of your choice, and explain why in your design doc.
4. **Let agents call at least 2 tools**, from the Startup Toolkit or your own. The LLM should decide when to call them; don't call them yourself.
5. **Produce a `launch_kit` dictionary** that passes the Mission Check (section 7).
6. **Run top to bottom** after *Restart Kernel → Run All*.

### 🎨 You are free to…
- pick an idea from the deck **or invent your own**,
- design any architecture: a pipeline, a graph, a hierarchy, a debate, a swarm…,
- choose the agents, roles, prompts, tools and patterns, and how many iterations they get,
- add extra sections to the launch kit (logo prompt, landing page, customer interviews, risk register…).

### 🤝 Fair play
- The agents must generate the content. No hard-coding the answers!
- Keep a full run under **~50 LLM calls**, because we all share one Azure OpenAI quota.
- Never print, hard-code or commit API keys.

## 📦 The Launch Kit (your deliverable)

Your system must produce a Python `dict` called **`launch_kit`** with the keys below. Values can be strings, lists, dicts or Pydantic models; the format inside each section is up to you.

| Key | What it should contain |
|---|---|
| `startup_name` | A catchy name you've checked is available |
| `one_liner` | A one-sentence elevator pitch |
| `target_customers` | Who exactly it's for (personas/segments) |
| `market_insights` | Market size, growth, trends and risks |
| `competitors` | Main competitors and how you'll beat them |
| `mvp_features` | A prioritised list of MVP features |
| `pricing` | Pricing model and price points |
| `unit_economics` | LTV, CAC, LTV/CAC and payback, calculated with the tool |
| `marketing_campaign` | Slogan, launch channels and a few social posts |
| `investor_pitch` | Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script) |
| `investor_feedback` | What the investor panel said (scores, verdicts, what changed) |

Extra keys are welcome. `show_launch_kit()` renders them too, and it will even display a `landing_page_html` section as a web page. 🌐

```python
from langchain.agents import create_agent
```

```python
import litellm
from google.adk.agents import LlmAgent, LoopAgent, ParallelAgent, SequentialAgent

from google.adk.runners import InMemoryRunner

litellm.suppress_debug_info = True
model = LiteLlm(model=f"openai/{AZURE_OPENAI_DEPLOYMENT}", api_base=AZURE_OPENAI_ENDPOINT,
                api_key=os.environ["AZURE_OPENAI_API_KEY"])

events = await runner.run_debug("Summarise the pets market.")
```

**Useful:** the workflow agents `SequentialAgent`, `ParallelAgent` and `LoopAgent`. `output_key` saves an agent's answer in session state, and `{market_report}` in another agent's instruction reads it back. `sub_agents=[...]` lets the LLM delegate.

<details>
<summary><b>👥 CrewAI</b> (use a fresh environment)</summary>

```python
from crewai import LLM, Agent, Crew, Process, Task
from crewai.tools import tool

llm = LLM(model=f"openai/{AZURE_OPENAI_DEPLOYMENT}", base_url=AZURE_OPENAI_ENDPOINT,
          api_key=os.environ["AZURE_OPENAI_API_KEY"])
market_tool = tool("Market Data")(get_market_data)

analyst = Agent(role="Market Analyst", goal="Size up startup markets", backstory="You love data.",
                tools=[market_tool], llm=llm)
task = Task(description="Summarise the {sector} market.", expected_output="3 bullet points", agent=analyst)
crew = Crew(agents=[analyst], tasks=[task], process=Process.sequential)
result = await crew.kickoff_async(inputs={"sector": "pets"})  # in notebooks, always use kickoff_async
print(result)
```

**Useful:** `context=[...]` chains tasks, `async_execution=True` runs tasks in parallel, `Process.hierarchical` with `manager_llm=llm` adds a manager agent, and `output_pydantic=MyPydanticModel` gives structured outputs.
</details>

````python
# 🧰 The Startup Toolkit: fictional mock data + framework-agnostic tools.
# Every tool is a plain Python function with type hints and a docstring,
# so you can wrap it with any framework's tool helper (see the cheat sheet).
import hashlib
import json
import random
import re

from IPython.display import HTML, Markdown, display

# ---------------------------------------------------------------------------
# 🃏 The Wild Idea Deck
# ---------------------------------------------------------------------------
WILD_IDEA_DECK = [
    {"id": 1, "sector": "home",
     "idea": "An on-demand app that matches travelling plant owners with vetted 'plant sitters' who water, talk to, and send selfies of your plants."},
    {"id": 2, "sector": "entertainment",
     "idea": "An AI dream journal that turns parents' dreams into personalised bedtime stories for their kids."},
    {"id": 3, "sector": "pets",
     "idea": "Noise-cancelling headphones for dogs that switch on automatically during fireworks and thunderstorms."},
    {"id": 4, "sector": "food",
     "idea": "A smart lunchbox and companion app that turns eating vegetables into a game for picky kids."},
    {"id": 5, "sector": "education",
     "idea": "A language-learning app where you practise by debating AI versions of historical figures."},
    {"id": 6, "sector": "social",
     "idea": "'Grandparents-as-a-Service': matching retirees with busy families for storytelling, homework help and cooking lessons."},
    {"id": 7, "sector": "fitness",
     "idea": "Gym equipment that turns your workout into electricity and credits it to your home energy bill."},
    {"id": 8, "sector": "sustainability",
     "idea": "Carbon-negative coffee pods grown from mushroom mycelium that you can plant in your garden after use."},
    {"id": 9, "sector": "mobility",
     "idea": "A marketplace where office buildings rent their empty parking spots to food trucks at lunchtime."},
    {"id": 10, "sector": "entertainment",
     "idea": "A dating app for board-game lovers where every first date is a cooperative game night."},
]

# ---------------------------------------------------------------------------
# 📊 Fictional market data (numbers are made up for this exercise!)
# ---------------------------------------------------------------------------
MARKET_DATA = {
    "home": {"market_size_usd_bn": 48, "annual_growth_pct": 7.5, "avg_spend_usd_per_month": 25,
             "trends": ["Urban apartment gardening boom", "Gig-economy home services", "Smart-home sensors"],
             "risks": ["Trust & safety for in-home visits", "Seasonal demand (holidays)", "Low willingness to pay"],
             "benchmarks": {"typical_cac_usd": 30, "typical_monthly_churn_pct": 8}},
    "entertainment": {"market_size_usd_bn": 210, "annual_growth_pct": 9.2, "avg_spend_usd_per_month": 12,
                      "trends": ["Generative-AI personalised content", "Niche communities over mass platforms", "Offline-first experiences"],
                      "risks": ["Content moderation", "Subscription fatigue", "Fast-moving copycats"],
                      "benchmarks": {"typical_cac_usd": 18, "typical_monthly_churn_pct": 9}},
    "pets": {"market_size_usd_bn": 320, "annual_growth_pct": 6.1, "avg_spend_usd_per_month": 120,
             "trends": ["Pet humanisation ('fur babies')", "Premium wellness & anxiety products", "Subscription boxes"],
             "risks": ["Animal-welfare and vet claims", "Hardware returns", "Seasonal spikes (fireworks season)"],
             "benchmarks": {"typical_cac_usd": 35, "typical_monthly_churn_pct": 6}},
    "food": {"market_size_usd_bn": 95, "annual_growth_pct": 5.4, "avg_spend_usd_per_month": 40,
             "trends": ["Healthy eating for kids", "Gamified nutrition", "Connected kitchen devices"],
             "risks": ["Food-safety regulations", "Parents' screen-time concerns", "Hardware margins"],
             "benchmarks": {"typical_cac_usd": 40, "typical_monthly_churn_pct": 7}},
    "education": {"market_size_usd_bn": 165, "annual_growth_pct": 12.8, "avg_spend_usd_per_month": 15,
                  "trends": ["AI tutors and conversation practice", "Gamification", "Micro-learning"],
                  "risks": ["Low completion rates", "Historical accuracy & bias", "Free alternatives"],
                  "benchmarks": {"typical_cac_usd": 22, "typical_monthly_churn_pct": 10}},
    "social": {"market_size_usd_bn": 60, "annual_growth_pct": 8.0, "avg_spend_usd_per_month": 80,
               "trends": ["Loneliness economy", "Intergenerational connection", "Care marketplaces"],
               "risks": ["Safeguarding & background checks", "Two-sided marketplace cold start", "Insurance"],
               "benchmarks": {"typical_cac_usd": 55, "typical_monthly_churn_pct": 5}},
    "fitness": {"market_size_usd_bn": 105, "annual_growth_pct": 6.9, "avg_spend_usd_per_month": 45,
                "trends": ["Gamified workouts", "Sustainability-minded consumers", "Connected equipment"],
                "risks": ["Expensive hardware", "Energy regulations", "Motivation drop-off after January"],
                "benchmarks": {"typical_cac_usd": 60, "typical_monthly_churn_pct": 6}},
    "sustainability": {"market_size_usd_bn": 140, "annual_growth_pct": 14.5, "avg_spend_usd_per_month": 30,
                       "trends": ["Carbon-negative products", "Bio-materials (mycelium, algae)", "Circular economy"],
                       "risks": ["Greenwashing accusations", "Manufacturing scale-up", "Premium pricing"],
                       "benchmarks": {"typical_cac_usd": 45, "typical_monthly_churn_pct": 4}},
    "mobility": {"market_size_usd_bn": 75, "annual_growth_pct": 10.3, "avg_spend_usd_per_month": 200,
                 "trends": ["Under-used urban space", "Food-truck culture", "B2B marketplaces"],
                 "risks": ["City permits & zoning", "Liability", "Chicken-and-egg marketplace dynamics"],
                 "benchmarks": {"typical_cac_usd": 150, "typical_monthly_churn_pct": 3}},
}

# ---------------------------------------------------------------------------
# 🏢 Fictional competitor database
# ---------------------------------------------------------------------------
COMPETITORS = [
    {"name": "Leafy Nanny", "sector": "home", "tagline": "Plant-care tips from a chatbot", "keywords": ["plants", "plant care", "gardening"], "pricing": "$4.99/month", "weakness": "No in-person service"},
    {"name": "GreenThumb Robotics", "sector": "home", "tagline": "Self-watering smart pots", "keywords": ["plants", "smart pot", "watering", "iot"], "pricing": "$89 one-time", "weakness": "Only works with their own pots"},
    {"name": "SnoozeStory", "sector": "entertainment", "tagline": "AI bedtime stories on demand", "keywords": ["bedtime", "stories", "kids", "ai", "parents"], "pricing": "$7.99/month", "weakness": "Generic stories, nothing from the family's own life"},
    {"name": "DreamLog", "sector": "entertainment", "tagline": "Track and analyse your dreams", "keywords": ["dream", "journal", "sleep"], "pricing": "Free / $3.99 pro", "weakness": "Analysis only, no creative output"},
    {"name": "Roll for Love", "sector": "entertainment", "tagline": "Dating for video gamers", "keywords": ["dating", "games", "gamers"], "pricing": "Freemium, $14.99/month", "weakness": "Video games only, no tabletop community"},
    {"name": "StormHug", "sector": "pets", "tagline": "Anxiety-calming vests for dogs", "keywords": ["dog", "anxiety", "fireworks", "thunder"], "pricing": "$45 one-time", "weakness": "Doesn't address noise itself"},
    {"name": "Barkly Box", "sector": "pets", "tagline": "Monthly toy & treat box for dogs", "keywords": ["dog", "subscription", "toys", "treats"], "pricing": "$29/month", "weakness": "Commodity product, high churn"},
    {"name": "VeggieQuest", "sector": "food", "tagline": "Nutrition game app for kids", "keywords": ["kids", "vegetables", "game", "nutrition"], "pricing": "$5.99/month", "weakness": "Parents worry about extra screen time"},
    {"name": "LunchBotics", "sector": "food", "tagline": "Temperature-controlled smart bento", "keywords": ["lunchbox", "smart", "kids", "bento"], "pricing": "$129 one-time", "weakness": "Expensive, no engagement features"},
    {"name": "Lingo Legends", "sector": "education", "tagline": "Learn languages through quests", "keywords": ["language", "learning", "gamification"], "pricing": "Freemium, $9.99/month", "weakness": "Scripted dialogues, no real conversation"},
    {"name": "ChronoChat", "sector": "education", "tagline": "Chat with historical figures", "keywords": ["history", "historical figures", "chat", "ai"], "pricing": "$9.99/month", "weakness": "Trivia-focused, not designed for language practice"},
    {"name": "GrandBuddies", "sector": "social", "tagline": "Companion visits for seniors", "keywords": ["seniors", "retirees", "companionship", "elderly"], "pricing": "$25/hour", "weakness": "Treats seniors as recipients, not contributors"},
    {"name": "Village Nanny", "sector": "social", "tagline": "Babysitter marketplace", "keywords": ["childcare", "families", "babysitting", "marketplace"], "pricing": "15% booking fee", "weakness": "Transactional, no long-term relationships"},
    {"name": "PowerPedal", "sector": "fitness", "tagline": "Bike desks that charge your devices", "keywords": ["electricity", "energy", "bike", "workout"], "pricing": "$399 one-time", "weakness": "Tiny energy output, novelty factor"},
    {"name": "FitCoin Gym", "sector": "fitness", "tagline": "Gym that rewards workouts with points", "keywords": ["gym", "rewards", "workout", "points"], "pricing": "$35/month", "weakness": "Points have little real-world value"},
    {"name": "ShroomPack", "sector": "sustainability", "tagline": "Mycelium packaging for brands", "keywords": ["mycelium", "mushroom", "packaging", "b2b"], "pricing": "B2B contracts", "weakness": "No consumer brand"},
    {"name": "EcoPod Co.", "sector": "sustainability", "tagline": "Compostable coffee pods", "keywords": ["coffee", "pods", "compostable", "capsules"], "pricing": "$0.60/pod", "weakness": "Compostable is not carbon-negative"},
    {"name": "SpotSwap", "sector": "mobility", "tagline": "Peer-to-peer parking rental", "keywords": ["parking", "rental", "marketplace", "cars"], "pricing": "20% commission", "weakness": "Cars only, no commercial vehicles"},
    {"name": "TruckTrack", "sector": "mobility", "tagline": "Find food trucks near you", "keywords": ["food truck", "locator", "lunch"], "pricing": "Free (ads)", "weakness": "Consumer app only, doesn't solve where trucks can park"},
]

# ---------------------------------------------------------------------------
# 🦈 The Demo Day investor panel (fictional)
# ---------------------------------------------------------------------------
INVESTOR_PANEL = [
    {"name": "Ada Okafor", "nickname": "The Analyst", "fund": "Spreadsheet Capital",
     "loves": ["big markets", "healthy unit economics (LTV/CAC >= 3)", "defensible moats"],
     "hates": ["hand-wavy numbers", "tiny niches"],
     "catchphrase": "Show me the numbers."},
    {"name": "Rex Moon", "nickname": "The Visionary", "fund": "Moonshot Ventures",
     "loves": ["bold visions", "great storytelling", "creating new categories"],
     "hates": ["me-too products", "boring names"],
     "catchphrase": "Will this change how people live?"},
    {"name": "Priya Shah", "nickname": "The Operator", "fund": "Shipyard Partners",
     "loves": ["a tight MVP", "a clear go-to-market plan", "realistic milestones"],
     "hates": ["feature bloat", "vague target customers"],
     "catchphrase": "What ships in the first 90 days?"},
]

_TAKEN_NAMES = {"nova", "spark", "zen", "pulse", "flow", "bloom", "apex", "orbit", "nexus", "hive", "leaf", "dream"}


def draw_wild_idea(seed: int | None = None) -> dict:
    """Draw a random idea card from the Wild Idea Deck (pass a seed to make it repeatable)."""
    return random.Random(seed).choice(WILD_IDEA_DECK)


# ---------------------------------------------------------------------------
# 🔧 Tools (give these to your agents!)
# ---------------------------------------------------------------------------
def get_market_data(sector: str) -> str:
    """Get market size, growth, trends, risks and benchmark CAC/churn for a market sector.

    Args:
        sector: Market sector, one of: home, entertainment, pets, food, education, social, fitness, sustainability, mobility.

    Returns:
        JSON string with the (fictional) market data for the sector.
    """
    key = sector.strip().lower()
    if key not in MARKET_DATA:
        return json.dumps({"error": f"Unknown sector '{sector}'.", "available_sectors": sorted(MARKET_DATA)})
    return json.dumps({"sector": key, **MARKET_DATA[key]})


def search_competitors(keyword: str) -> str:
    """Search the competitor database for companies matching a keyword (e.g. 'dog', 'parking', 'coffee').

    Args:
        keyword: A single keyword or short phrase describing the product or customer.

    Returns:
        JSON string with up to 5 matching competitors, including their pricing and main weakness.
    """
    needle = keyword.strip().lower()
    matches = [
        c for c in COMPETITORS
        if needle and (needle in c["name"].lower() or needle in c["tagline"].lower()
                       or needle == c["sector"] or any(needle in k for k in c["keywords"]))
    ]
    if not matches:
        return json.dumps({"keyword": keyword, "competitors": [], "note": "No direct competitors found. Blue ocean, or no market?"})
    return json.dumps({"keyword": keyword, "competitors": matches[:5]})


def check_name_availability(name: str) -> str:
    """Check whether a startup name is available as a .com/.ai domain and whether it has trademark risk.

    Args:
        name: The proposed startup name, e.g. 'BarkBuds'.

    Returns:
        JSON string with domain availability, trademark risk and a verdict.
    """
    slug = re.sub(r"[^a-z0-9]", "", name.lower())
    if not slug:
        return json.dumps({"name": name, "error": "Name must contain letters or digits."})
    digest = int(hashlib.sha256(slug.encode()).hexdigest(), 16)
    clash = any(taken in slug for taken in _TAKEN_NAMES)
    com_free = not clash and digest % 3 != 0
    ai_free = not clash and digest % 4 != 0
    risk = "high" if clash else ("medium" if digest % 7 == 0 else "low")
    usable = (com_free or ai_free) and risk != "high"
    return json.dumps({
        "name": name,
        "domain_com": f"{slug}.com is " + ("available" if com_free else "taken"),
        "domain_ai": f"{slug}.ai is " + ("available" if ai_free else "taken"),
        "trademark_risk": risk,
        "verdict": "Looks usable!" if usable else "Pick another name.",
    })


def estimate_unit_economics(price_per_month_usd: float, cost_to_serve_per_month_usd: float,
                            acquisition_cost_usd: float, monthly_churn_pct: float) -> str:
    """Calculate customer lifetime value (LTV), LTV/CAC ratio and CAC payback period.

    For one-off purchases, spread the price over the expected product life to get a monthly figure.

    Args:
        price_per_month_usd: Average monthly revenue per customer in USD.
        cost_to_serve_per_month_usd: Average monthly cost to serve one customer in USD.
        acquisition_cost_usd: Customer acquisition cost (CAC) in USD.
        monthly_churn_pct: Percentage of customers lost each month, e.g. 5 for 5%.

    Returns:
        JSON string with gross margin, LTV, LTV/CAC, payback months and a verdict.
    """
    margin = price_per_month_usd - cost_to_serve_per_month_usd
    churn = max(monthly_churn_pct, 0.1) / 100
    lifetime_months = 1 / churn
    ltv = margin * lifetime_months
    ltv_cac = ltv / acquisition_cost_usd if acquisition_cost_usd > 0 else float("inf")
    payback = acquisition_cost_usd / margin if margin > 0 else float("inf")
    if margin <= 0:
        verdict = "Negative margin: you lose money on every customer."
    elif ltv_cac >= 3:
        verdict = "Healthy: investors like LTV/CAC >= 3."
    elif ltv_cac >= 1:
        verdict = "Risky: profitable per customer, but growth will be expensive."
    else:
        verdict = "Unsustainable: each customer costs more than they bring in."
    return json.dumps({
        "gross_margin_per_month_usd": round(margin, 2),
        "gross_margin_pct": round(100 * margin / price_per_month_usd, 1) if price_per_month_usd else None,
        "expected_lifetime_months": round(lifetime_months, 1),
        "ltv_usd": round(ltv, 2),
        "ltv_to_cac": round(ltv_cac, 2) if ltv_cac != float("inf") else "infinite",
        "cac_payback_months": round(payback, 1) if payback != float("inf") else "never",
        "verdict": verdict,
    })


def get_investor_panel() -> str:
    """Get the profiles of the Demo Day investors: what each one loves, hates and asks about.

    Returns:
        JSON string with the investor profiles.
    """
    return json.dumps(INVESTOR_PANEL)


# ---------------------------------------------------------------------------
# 📦 Launch Kit contract, checker and viewer
# ---------------------------------------------------------------------------
LAUNCH_KIT_SPEC = {
    "startup_name": "A catchy name you've checked is available",
    "one_liner": "A one-sentence elevator pitch",
    "target_customers": "Who exactly it's for (personas/segments)",
    "market_insights": "Market size, growth, trends and risks",
    "competitors": "Main competitors and how you'll beat them",
    "mvp_features": "A prioritised list of MVP features",
    "pricing": "Pricing model and price points",
    "unit_economics": "LTV, CAC, LTV/CAC and payback, calculated with the tool",
    "marketing_campaign": "Slogan, launch channels and a few social posts",
    "investor_pitch": "Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script)",
    "investor_feedback": "What the investor panel said (scores, verdicts, what changed)",
}


def _to_plain(value):
    """Convert Pydantic models (and containers of them) into plain Python data."""
    if hasattr(value, "model_dump"):
        return _to_plain(value.model_dump())
    if isinstance(value, dict):
        return {k: _to_plain(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [_to_plain(v) for v in value]
    return value


def _is_filled(value) -> bool:
    if value is None:
        return False
    if isinstance(value, str):
        return bool(value.strip())
    if isinstance(value, (list, tuple, dict, set)):
        return len(value) > 0
    return bool(str(value).strip())


def check_launch_kit(kit) -> bool:
    """Mission control check: verify that the launch kit contains every required section."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict):
        print(f"❌ launch_kit must be a dict (or a Pydantic model), got {type(kit).__name__}.")
        return False
    passed = 0
    for key, description in LAUNCH_KIT_SPEC.items():
        ok = _is_filled(kit.get(key))
        passed += ok
        print(f"{'✅' if ok else '❌'} {key:<20}" + ("" if ok else f" <- missing: {description}"))
    extras = [k for k in kit if k not in LAUNCH_KIT_SPEC]
    if extras:
        print(f"\n✨ Extra sections (nice!): {', '.join(extras)}")
    print(f"\nLaunch readiness: {passed}/{len(LAUNCH_KIT_SPEC)}")
    if passed == len(LAUNCH_KIT_SPEC):
        print("🚀 All systems go! Your startup is ready for Demo Day.")
    return passed == len(LAUNCH_KIT_SPEC)


def _render(value, indent: int = 0) -> str:
    pad = "  " * indent
    if isinstance(value, dict):
        lines = []
        for k, v in value.items():
            if isinstance(v, (dict, list)):
                lines.append(f"{pad}- **{k}**:\n{_render(v, indent + 1)}")
            else:
                lines.append(f"{pad}- **{k}**: {v}")
        return "\n".join(lines)
    if isinstance(value, list):
        return "\n".join(
            f"{pad}-\n{_render(v, indent + 1)}" if isinstance(v, (dict, list)) else f"{pad}- {v}" for v in value
        )
    return f"{pad}{value}"


def show_launch_kit(kit) -> None:
    """Render the launch kit as a nicely formatted Markdown report."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict) or not kit:
        print("Nothing to show yet. Build your launch_kit first!")
        return
    md = [f"# 🚀 {kit.get('startup_name', 'Unnamed Startup')}"]
    if kit.get("one_liner"):
        md.append(f"> {kit['one_liner']}")
    for key, value in kit.items():
        if key in ("startup_name", "one_liner", "landing_page_html") or not _is_filled(value):
            continue
        md.append(f"## {key.replace('_', ' ').title()}\n\n{_render(value)}")
    display(Markdown("\n\n".join(md)))
    if kit.get("landing_page_html"):
        display(HTML(kit["landing_page_html"]))


print(f"🧰 Startup Toolkit loaded: {len(WILD_IDEA_DECK)} wild ideas, {len(MARKET_DATA)} sectors, "
      f"{len(COMPETITORS)} competitors, {len(INVESTOR_PANEL)} investors.")
````

### 📝 My Design Doc

**Framework:** LangGraph. My design is a graph: two things start at the same time, they join, the work goes through a chain of steps, and it ends in a loop that stops on a score. LangGraph lets me write each of those directly as nodes and edges, and I keep the whole state in my hands, so the real tool numbers end up in the launch kit instead of the LLM's own maths. The agents that use tools are LangChain `create_agent` agents running inside the graph nodes.

**My idea:** My own: Dish Duel. A clip-on camera watches the sink, and every flatmate gets their own colour of plates and mugs. When a dish sits there too long, the app posts a dramatic "MOST WANTED" poster in the flat group chat, and whoever washes the most is crowned "Dish Champion of the week". Before picking it I searched: people keep building DIY versions (a hackspace dish detector, hackathon apps like DishWatcher), which proves the demand, but I couldn't find a real product. Some of those versions use face recognition, which is creepy, so ours only looks at the sink and lets the plate colour tell on you. Sector: home.

**Architecture:** a graph pipeline. A parallel fan-out for market research and branding, a fan-in to the product lead, a sequential chain through marketing to the pitch, and an evaluator-optimizer loop where three investor agents score the pitch until the average hits 8 or three rounds pass. I chose a fixed graph over a "CEO" orchestrator because the order of the work is already clear (you can't price before you know the market), so letting an LLM plan it would only add cost and unpredictability. The only open question is quality, and that's what the loop handles.

**Agent team**

| Agent | Role / responsibility | Tools | Produces |
|---|---|---|---|
| Market Analyst | Pulls the market data, looks up competitors with a few keywords, writes personas | `get_market_data`, `search_competitors` | `market_insights`, `competitors`, `target_customers` |
| Brand Strategist | Brainstorms names, checks every one, and only keeps one the tool says is usable | `check_name_availability` | `startup_name`, `one_liner`, brand voice |
| Product & Finance Lead | Plans a 90-day MVP and pricing, runs the numbers, re-prices if LTV/CAC < 3 | `estimate_unit_economics` | `mvp_features`, `pricing`, `unit_economics` |
| Growth Marketer | Slogan, channels, social posts and the landing page | none | `marketing_campaign`, `landing_page_html` |
| Pitch Writer | 6-slide deck + ~200-word script, rewrites it from investor feedback | none | `investor_pitch` |
| Investor panel (Ada, Rex, Priya) | One agent per investor, built from `get_investor_panel()`. Each scores 1-10 and names one must-fix | none | `investor_feedback` |

**Patterns I use (where and why)**
- **Parallelization:** market research and branding don't depend on each other, so they run at the same time. The three investors also score in parallel. Faster, and the investors can't influence each other.
- **Prompt chaining:** product → marketing → pitch. Each step gets the output of the previous ones, so the pitch uses the real price, the real LTV/CAC and the real slogan.
- **Evaluator-optimizer (Shark Tank loop):** the panel scores the pitch, the pitch writer fixes the must-fix points, and we go again. It stops when the average is ≥ 8 or after 3 rounds.
- **Tool use / ReAct:** the agents decide which tools to call and how often. The brand agent keeps checking names until one passes, and the product lead re-runs the economics if the ratio is bad. That matters here because the kit is hardware plus a subscription, so the agent has to spread the hardware cost.
- Small extra: an optional **human-in-the-loop** veto on the name (`FOUNDER_MODE = True`).

**Workflow sketch**
```mermaid
flowchart LR
    idea([idea]) --> MR[Market Analyst]
    idea --> BS[Brand Strategist]
    MR --> PL[Product & Finance Lead]
    BS --> PL
    PL --> MK[Growth Marketer]

    PW --> ST{{Shark Tank: Ada, Rex, Priya}}
    ST -- avg < 8 and round < 3 --> PW
    ST -- avg >= 8 or 3 rounds --> kit([launch_kit])
```

**Stop conditions and safety limits**
- Shark Tank: max 3 rounds, or stop as soon as the average is ≥ 8
- Every agent has `recursion_limit=14`, so a tool loop can't run forever
- The brand agent gets max 2 attempts. The product lead is told max 3 economics runs
- `unit_economics` is taken from the agent's last *actual* tool call, not from what the LLM writes, so the numbers can't be made up
- Duplicate competitors (the same company found by two searches) are removed in code
- All structured outputs are Pydantic models, so a bad format fails loudly instead of quietly breaking the kit
- Expected budget: about 16-30 LLM calls for the main run, plus ~12-15 for the pitch battle (the cap is ~50)

- 📐 **Structured outputs** (Pydantic models) make assembling the `launch_kit` much easier.

````python
# 5.1 Model, tools, schemas and a small tracker
import json
import operator
import time
from typing import Annotated, Literal, TypedDict

from langchain.agents import create_agent
from langchain.agents.structured_output import ToolStrategy
from langchain_core.callbacks import BaseCallbackHandler
from langchain_core.messages import ToolMessage
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.graph import END, START, StateGraph
from pydantic import BaseModel, Field

# no temperature / max_tokens here, gpt-5 models reject them
llm = ChatOpenAI(model=AZURE_OPENAI_DEPLOYMENT, base_url=AZURE_OPENAI_ENDPOINT,
                 api_key=os.environ["OPENAI_API_KEY"])

# wrap the toolkit functions so the agents can decide when to call them
market_tool = tool(get_market_data)
competitor_tool = tool(search_competitors)
name_tool = tool(check_name_availability)
economics_tool = tool(estimate_unit_economics)


# ---------- structured outputs (each agent fills one of these) ----------
class Persona(BaseModel):
    name: str = Field(description="Short persona label, e.g. 'Tired night-shift nurse'")
    who: str = Field(description="Who they are, in one sentence")
    pain_point: str = Field(description="The problem we solve for them")


class Competitor(BaseModel):
    name: str
    pricing: str
    weakness: str
    how_we_win: str = Field(description="One sentence on how we beat them")


class MarketReport(BaseModel):
    market_size_usd_bn: float
    annual_growth_pct: float
    avg_spend_usd_per_month: float
    trends: list[str]
    risks: list[str]
    benchmark_cac_usd: float
    benchmark_monthly_churn_pct: float
    so_what: str = Field(description="2-3 sentences: what this market means for our idea")
    target_customers: list[Persona] = Field(description="2-3 specific personas")
    competitors: list[Competitor]


class NameCandidate(BaseModel):
    name: str
    tool_verdict: str = Field(description="What check_name_availability said about it")


class BrandKit(BaseModel):
    startup_name: str = Field(description="The chosen name. The tool must have said 'Looks usable!'")
    one_liner: str = Field(description="One-sentence elevator pitch, under 25 words")
    brand_voice: str = Field(description="How the brand talks, in one sentence")
    names_considered: list[NameCandidate]


class Feature(BaseModel):
    priority: int = Field(description="1 = build first")
    name: str
    what_it_does: str


class PricePlan(BaseModel):
    plan: str
    price: str
    includes: str


class ProductPlan(BaseModel):
    mvp_features: list[Feature] = Field(description="4-6 features, highest priority first")
    first_90_days: list[str] = Field(description="3-4 milestones, one per month-ish")
    pricing_model: str
    price_points: list[PricePlan]
    economics_reasoning: str = Field(description="Why you picked those tool inputs and what you changed between runs")


class SocialPost(BaseModel):
    platform: str
    post: str


class Campaign(BaseModel):
    slogan: str
    launch_channels: list[str]
    social_posts: list[SocialPost] = Field(description="3 posts for different platforms")
    landing_page_html: str = Field(description="One self-contained HTML page with inline CSS, no scripts, no external images")


class Slide(BaseModel):
    title: str
    points: list[str]


class Pitch(BaseModel):
    slides: list[Slide] = Field(description="Exactly 6 slides")
    script: str = Field(description="About 200 words, spoken out loud at Demo Day")
    what_changed: str = Field(description="What you changed since the last round and why ('first draft' in round 1)")


class InvestorVerdict(BaseModel):
    score: int = Field(ge=1, le=10, description="1-10, be tough: 8+ means you would really invest")
    decision: Literal["invest", "maybe", "pass"]
    liked: str
    concerns: str
    must_fix: str = Field(description="The one thing the founders must fix before you'd invest")


# ---------- tracker: LLM calls, tokens and time per stage, plus every tool call ----------
class StudioTracker(BaseCallbackHandler):
    def __init__(self):
        self.reset()

    def reset(self):
        self.stats = {}
        self._runs = {}  # run_id -> stage, so tokens go to the right stage

    def stage(self, name):
        return self.stats.setdefault(name, {"llm_calls": 0, "tokens": 0, "seconds": 0.0})

    def on_chat_model_start(self, serialized, messages, *, run_id, metadata=None, **kwargs):
        name = (metadata or {}).get("stage", "other")
        self._runs[run_id] = name
        self.stage(name)["llm_calls"] += 1

    def on_llm_end(self, response, *, run_id, **kwargs):
        name = self._runs.pop(run_id, "other")
        for gens in response.generations:
            for g in gens:
                usage = getattr(getattr(g, "message", None), "usage_metadata", None) or {}
                self.stage(name)["tokens"] += usage.get("total_tokens", 0)

    def on_tool_start(self, serialized, input_str, *, metadata=None, **kwargs):
        print(f"   🔧 [{(metadata or {}).get('stage', '?')}] {serialized.get('name')}({input_str})")

    def summary(self):
        rows = {k: {**v, "seconds": round(v["seconds"], 1)} for k, v in self.stats.items()}
        rows["TOTAL"] = {key: round(sum(v[key] for v in self.stats.values()), 1)
                         for key in ("llm_calls", "tokens", "seconds")}
        return rows


tracker = StudioTracker()


def run_config(stage):
    # every agent/LLM call gets its stage name, so the tracker knows where it came from
    return {"callbacks": [tracker], "metadata": {"stage": stage}, "recursion_limit": 14}


def tool_runs(messages, tool_name):
    """Pull out (args, result) for every time an agent called `tool_name`."""
    args = {tc["id"]: tc["args"] for m in messages for tc in (getattr(m, "tool_calls", None) or [])}
    runs = []
    for m in messages:
        if isinstance(m, ToolMessage) and m.name == tool_name:
            try:
                runs.append((args.get(m.tool_call_id, {}), json.loads(m.content)))
            except (TypeError, json.JSONDecodeError):
                pass  # tool errored, skip it
    return runs


print("✅ Model, 4 tools and schemas ready")
````

````python
# 5.2 The agents
FOUNDER_MODE = False # True = you approve/veto the name with input() 
MAX_ROUNDS = 3        # shark tank rounds
TARGET_SCORE = 8      # average score we need to stop early
TONE = "Funny and dramatic like a wanted poster or a sports final. Tease the dishes, never the people."

market_analyst = create_agent(
    llm,
    tools=[market_tool, competitor_tool],
    response_format=ToolStrategy(MarketReport),
    system_prompt=(
        "You are the market analyst at a venture studio. "
        "First call get_market_data for the sector you're given. "
        "Then call search_competitors 2-3 times with different short keywords from the idea (e.g. 'smart', 'iot', 'home'). "
        "Only use numbers that came back from the tools, never invent market figures. "
        "Write 2-3 specific target personas (real flat-shares, not 'young people in general') "
        "and say in one line how we beat each competitor, using its weakness."
    ),
)

brand_strategist = create_agent(
    llm,
    tools=[name_tool],
    response_format=ToolStrategy(BrandKit),
    system_prompt=(
        "You are a brand strategist who hates boring names. "
        "Brainstorm 4 short, catchy, original names for the startup, then check EACH one with check_name_availability. "
        "You may only pick a name the tool called 'Looks usable!'. If none pass, invent more and check those (max 8 checks). "
        "Then write a one-liner under 25 words and describe the brand voice."
    ),
)

product_lead = create_agent(
    llm,
    tools=[economics_tool],
    response_format=ToolStrategy(ProductPlan),
    system_prompt=(
        "You are the product and finance lead. Plan a tight MVP that can ship in 90 days (no feature bloat) "
        "and a pricing model. Then call estimate_unit_economics with your monthly price, a realistic monthly cost to "
        "serve (camera + plate kit hardware, app hosting, support), and the benchmark CAC and churn from the market report. "
        "Only change CAC or churn if you can say why. "
        "Spread the hardware kit price over its expected life (e.g. 24 months) and add a small subscription; "
        "one flat shares one kit, so price per flat. "
        "If LTV/CAC comes back under 3, change the pricing or plan and call the tool again (max 3 calls). "
        "Your final pricing must match your LAST tool call."
    ),
)

# these three just write, no tools needed, so a plain structured LLM is enough
marketer = llm.with_structured_output(Campaign, method="function_calling")
pitch_writer = llm.with_structured_output(Pitch, method="function_calling")
investor_judge = llm.with_structured_output(InvestorVerdict, method="function_calling")

# one investor "agent" per panel member: same model, different persona
INVESTORS = json.loads(get_investor_panel())


def investor_prompt(inv):
    return (f"You are {inv['name']} ('{inv['nickname']}') from {inv['fund']} at Demo Day. "
            f"You love: {', '.join(inv['loves'])}. You hate: {', '.join(inv['hates'])}. "
            f"Your catchphrase: '{inv['catchphrase']}'. "
            "This is a PRE-LAUNCH startup, so judge the idea, plan and numbers it has; don't demand data "
            "that can't exist yet (like real retention). Your must_fix should be something the founders can "
            "change in the pitch itself. Score honestly from 1-10; most pitches get 5-7.")


print(f"✅ 3 tool agents + marketer + pitch writer + {len(INVESTORS)} investors: "
      + ", ".join(i["name"] for i in INVESTORS))
````

````python
# 5.3 The workflow graph
from langchain_core.runnables.graph import NodeStyles


class StudioState(TypedDict, total=False):
    idea: str
    sector: str
    market: MarketReport
    brand: BrandKit
    name_check: dict
    product: ProductPlan
    unit_economics: dict
    campaign: Campaign
    pitch: Pitch
    round: int
    feedback: Annotated[list, operator.add]  # each shark tank round adds one entry


def timed(stage):
    """Print when a stage starts and add its wall time to the tracker."""
    def wrap(fn):
        def node(state):
            print(f"▶️  {stage}")
            start = time.perf_counter()
            out = fn(state, run_config(stage))
            tracker.stage(stage)["seconds"] += time.perf_counter() - start
            return out
        return node
    return wrap


def ask(text):
    return {"messages": [{"role": "user", "content": text}]}


def brief(state):
    """Short JSON summary of the startup so far, for the pitch writer and the investors."""
    m, b, p = state["market"], state["brand"], state["product"]
    return json.dumps({
        "name": b.startup_name, "one_liner": b.one_liner, "idea": state["idea"],
        "market": {"size_usd_bn": m.market_size_usd_bn, "growth_pct": m.annual_growth_pct, "risks": m.risks},
        "target_customers": [x.model_dump() for x in m.target_customers],
        "competitors": [x.model_dump() for x in m.competitors],
        "mvp_features": [x.model_dump() for x in p.mvp_features],
        "first_90_days": p.first_90_days,
        "pricing": [x.model_dump() for x in p.price_points],
        "unit_economics": state.get("unit_economics"),
    }, indent=1)


# --- stage 1 (parallel): market research + branding ---
@timed("market_research")
def market_research(state, config):
    res = market_analyst.invoke(ask(f"Idea: {state['idea']}\nSector: {state['sector']}"), config)
    report = res["structured_response"]
    seen = set()  # drop competitors that came back from more than one search
    report.competitors = [c for c in report.competitors if not (c.name in seen or seen.add(c.name))]
    return {"market": report}


@timed("brand_studio")
def brand_studio(state, config):
    extra = ""
    for attempt in range(2):  # max 2 tries, then we take what we have
        res = brand_strategist.invoke(ask(f"Idea: {state['idea']}{extra}"), config)
        brand = res["structured_response"]
        checks = {a.get("name", "").lower(): r for a, r in tool_runs(res["messages"], "check_name_availability")}
        check = checks.get(brand.startup_name.lower())
        if not check or check.get("verdict") != "Looks usable!":
            print(f"   ⚠️ '{brand.startup_name}' wasn't confirmed by the tool, trying again")
            extra = f"\n\nDon't use '{brand.startup_name}', the tool didn't confirm it. Check new names."
            continue
        if FOUNDER_MODE and input(f"   🧑‍💼 Founder: keep the name '{brand.startup_name}'? [y/n] ").strip().lower() == "n":
            extra = f"\n\nThe founder vetoed '{brand.startup_name}'. Come up with different names."
            continue
        break
    print(f"   ✅ name: {brand.startup_name}")
    return {"brand": brand, "name_check": check}


# --- stage 2: product + unit economics (needs both branches) ---
@timed("product_lead")
def product_and_numbers(state, config):
    m = state["market"]
    convo = ask(
        f"Startup: {state['brand'].startup_name} - {state['brand'].one_liner}\nIdea: {state['idea']}\n"
        f"Personas: {[x.model_dump() for x in m.target_customers]}\n"
        f"Competitor pricing: {[(c.name, c.pricing) for c in m.competitors]}\n"
        f"Benchmarks: CAC ${m.benchmark_cac_usd}, monthly churn {m.benchmark_monthly_churn_pct}%, "
        f"average spend ${m.avg_spend_usd_per_month}/month")
    for attempt in range(2):  # 1 retry max if the numbers are still bad
        res = product_lead.invoke(convo, config)
        runs = tool_runs(res["messages"], "estimate_unit_economics")
        if not runs:
            print("   ⚠️ the product lead never called estimate_unit_economics!")
            return {"product": res["structured_response"], "unit_economics": {}}
        ratio = runs[-1][1]["ltv_to_cac"]
        if ratio == "infinite" or ratio >= 3 or attempt == 1:
            break
        # the agent stopped with LTV/CAC under 3, so send it back with the tool's verdict
        print(f"   ↩️ LTV/CAC {ratio} is under 3, sending the product lead back to re-price")
        convo = {"messages": res["messages"] + [{"role": "user", "content":
                 f"Your last estimate_unit_economics run gave LTV/CAC {ratio}, which is under 3. Change the plan "
                 f"(price, plan structure or cost to serve, or CAC/churn only with a clear reason), call the tool "
                 f"again, and make your final pricing match that new run."}]}
    inputs, result = runs[-1]  # the final run is the one that counts
    economics = {"inputs": inputs, **result, "tool_runs": len(runs),
                 "history": [f"${a.get('price_per_month_usd')}/mo -> LTV/CAC {r['ltv_to_cac']}" for a, r in runs]}
    print(f"   ✅ LTV/CAC {result['ltv_to_cac']} after {len(runs)} tool run(s)")
    return {"product": res["structured_response"], "unit_economics": economics}

# --- stage 3: marketing ---
@timed("marketing")
def marketing(state, config):
    b, p = state["brand"], state["product"]
    campaign = marketer.invoke(
        "You are a playful growth marketer. Write a launch campaign: a slogan, 4-5 launch channels that fit "
        "these personas, and 3 social posts for different platforms. Also write a landing page as one self-contained "
        "HTML page (inline CSS, no JavaScript, no external images, under 120 lines) with a hero, 3 features, "
        f"pricing and a sign-up button. {TONE}\n\n"
        f"Brand voice: {b.brand_voice}\nPricing model: {p.pricing_model}\n\n{brief(state)}", config)
    return {"campaign": campaign}


# --- stage 4: pitch writer + shark tank loop ---
@timed("pitch_writer")
def write_pitch(state, config):
    rnd = state.get("round", 0)
    if rnd == 0:
        task = "Write the first Demo Day pitch."
    else:
        last = state["feedback"][-1]
        notes = "\n".join(f"- {v['investor']} ({v['score']}/10): {v['must_fix']}" for v in last["verdicts"])
        task = (f"Round {rnd} scored {last['average']}/10. Revise the pitch to fix this feedback. "
                f"Don't invent new numbers, use the ones in the summary.\n{notes}\n\n"
                f"Previous script:\n{state['pitch'].script}")
    pitch = pitch_writer.invoke(
        "You write Demo Day pitches: 6 slides (problem, solution, market, business model, go-to-market, ask) "
        f"and a ~200-word script. Use the real numbers below, especially LTV/CAC and payback. {TONE}\n\n"
        f"Slogan: {state['campaign'].slogan}\n{brief(state)}\n\n{task}", config)
    return {"pitch": pitch}


@timed("shark_tank")
def shark_tank(state, config):
    pitch = state["pitch"]
    deck = "\n".join(f"{i}. {s.title}: {'; '.join(s.points)}" for i, s in enumerate(pitch.slides, 1))
    prompts = [[{"role": "system", "content": investor_prompt(inv)},
                {"role": "user", "content": f"Slides:\n{deck}\n\nScript:\n{pitch.script}\n\nFact sheet:\n{brief(state)}"}]
               for inv in INVESTORS]
    verdicts = investor_judge.batch(prompts, config)  # all investors score at the same time
    rnd = state.get("round", 0) + 1
    rows = [{"investor": inv["name"], **v.model_dump()} for inv, v in zip(INVESTORS, verdicts)]
    avg = round(sum(v.score for v in verdicts) / len(verdicts), 2)
    for r in rows:
        print(f"   🦈 {r['investor']}: {r['score']}/10 ({r['decision']})")
    print(f"   📊 round {rnd} average: {avg}")
    return {"round": rnd, "feedback": [{"round": rnd, "average": avg,
                                        "pitch_changes": pitch.what_changed, "verdicts": rows}]}


def next_step(state):
    # stop when investors are happy or we've used all rounds
    if state["feedback"][-1]["average"] >= TARGET_SCORE or state["round"] >= MAX_ROUNDS:
        return END
    return "pitch_writer"


graph = StateGraph(StudioState)
graph.add_node("market_research", market_research)
graph.add_node("brand_studio", brand_studio)
graph.add_node("product_lead", product_and_numbers)
graph.add_node("marketing", marketing)
graph.add_node("pitch_writer", write_pitch)
graph.add_node("shark_tank", shark_tank)

graph.add_edge(START, "market_research")   # these two run in parallel
graph.add_edge(START, "brand_studio")
graph.add_edge(["market_research", "brand_studio"], "product_lead")  # waits for both
graph.add_edge("product_lead", "marketing")
graph.add_edge("marketing", "pitch_writer")
graph.add_edge("pitch_writer", "shark_tank")
graph.add_conditional_edges("shark_tank", next_step, ["pitch_writer", END])

studio = graph.compile()

# custom node colours so the text is readable in dark and light mode
node_style = NodeStyles(
    default="fill:#1d3557,stroke:#a8dadc,stroke-width:2px,color:#ffffff",  # normal nodes
    first="fill:#457b9d,stroke:#a8dadc,color:#ffffff",                     # __start__
    last="fill:#457b9d,stroke:#a8dadc,color:#ffffff",                      # __end__
)

try:
    display(Markdown(f"```mermaid\n{studio.get_graph().draw_mermaid(node_colors=node_style)}\n```"))
except Exception:
    print(studio.get_graph().draw_ascii())
````

Run your venture studio on `MY_IDEA` and collect the results into the `launch_kit` dictionary. You can fill it straight from your agents' outputs, or have a final "assembler" agent build it.

````python

tracker.reset()
start = time.perf_counter()
final = studio.invoke({"idea": MY_IDEA["idea"], "sector": MY_IDEA["sector"], "round": 0})
total_seconds = time.perf_counter() - start

market, brand, product = final["market"], final["brand"], final["product"]
campaign, pitch, rounds = final["campaign"], final["pitch"], final["feedback"]
last = rounds[-1]

launch_kit = {
    "startup_name": brand.startup_name,
    "one_liner": brand.one_liner,
    "target_customers": market.target_customers,
    "market_insights": {
        "market_size": f"${market.market_size_usd_bn}bn",
        "annual_growth": f"{market.annual_growth_pct}%",
        "trends": market.trends,
        "risks": market.risks,
        "what_it_means_for_us": market.so_what,
    },
    "competitors": market.competitors,
    "mvp_features": product.mvp_features,
    "pricing": {"model": product.pricing_model, "plans": product.price_points},
    "unit_economics": final["unit_economics"],  # straight from the tool, not the LLM's own math
    "marketing_campaign": {"slogan": campaign.slogan, "launch_channels": campaign.launch_channels,
                           "social_posts": campaign.social_posts},
    "investor_pitch": {"slides": pitch.slides, "script": pitch.script},
    "investor_feedback": {
        "rounds_played": len(rounds),
        "score_per_round": [r["average"] for r in rounds],
        "outcome": "🎉 Panel convinced" if last["average"] >= TARGET_SCORE else "Still not at 8, ran out of rounds",
        "final_verdicts": last["verdicts"],
        "what_changed_each_round": [f"Round {r['round']}: {r['pitch_changes']}" for r in rounds],
    },
    # extras
    "name_check": {"tool_result": final.get("name_check"), "names_considered": brand.names_considered},
    "first_90_days": product.first_90_days,
    "economics_reasoning": product.economics_reasoning,
    "run_stats": tracker.summary(),
    "landing_page_html": campaign.landing_page_html,
}

print(f"\n🏁 Done in {total_seconds:.0f}s with {tracker.summary()['TOTAL']['llm_calls']} LLM calls")
````

## 8. 🪞 Reflection

1. **Which pattern added the most value, and why?**
The Shark Tank loop. All three investors said the same thing: the story mixed a camera product with a software-only one. The pitch writer fixed the story, but it can't change the MVP or the price, so the pitch and the plan now disagree. Scores stayed flat (6.0 → 6.33 → 6.0). A loop only helps if the feedback reaches an agent that can fix it.

2. **Where did your agents struggle, and how did you fix it?**
The product lead sometimes said "I'm raising the price" and then stopped without checking. Now my code reads the real tool result and sends it back if LTV/CAC is under 3. This run it wasn't needed ($14, LTV/CAC 3.96 on the first try). I also remove duplicate competitors in code, keep the tone in one setting, and show the landing page in an iframe so its CSS doesn't break the notebook

3. **How many LLM calls does one run take, and how long?**
21 calls, about 42k tokens, 47 seconds. The Shark Tank uses more than half the tokens, because every investor reads the full pitch each round. The pitch battle added 15 calls (36 total), and our startup won 3-0 against VoidVan. Our idea was always "Startup A", so a fairer test would swap the order.

4. **If you rebuilt this in another framework, what would be easier or harder?**
In Google ADK the parallel part and the loop are built in, so there'd be less code. But my custom stop rule and the economics check are easier in LangGraph, because I control the whole state.

5. **What would need to change before real founders could use this?**
Real data. The competitor list is fake, so the agents compared a dish app to plant pots. While picking my idea I searched about 15 ideas online and all of them already existed in some form, so a real version needs a web-search tool. For Dish Duel itself, the camera should only see the sink, everyone in the flat has to agree to it, and there should be an easy way to pause the "most wanted" posts.

````python
# Bonus 6: Pitch battle, Dish Duel vs a rival idea, the investor panel picks the winner
RUN_BATTLE = True  # False = skip it (saves ~12-15 LLM calls)
RIVAL_IDEA = {  # a rival that's nothing like ours: B2B, real-world, unhinged
    "idea": "Scream Vans: soundproof vans park in empty office parking spots at lunch. Burnt-out employees book "
            "10 minutes to scream into the void, cry dramatically or blast their sad playlist, and get a monthly "
            "'Scream Wrapped'. Companies pay for it as a wellbeing perk.",
    "sector": "mobility",
}


class BattleVote(BaseModel):
    winner: Literal["A", "B"]
    reason: str = Field(description="One or two sentences on why")


def pitch_card(state):
    """What each investor sees about one startup in the battle."""
    eco = state["unit_economics"]
    return (f"{state['brand'].startup_name}: {state['brand'].one_liner}\n"
            f"Script: {state['pitch'].script}\n"
            f"LTV/CAC {eco.get('ltv_to_cac')}, payback {eco.get('cac_payback_months')} months, "
            f"panel score {state['feedback'][-1]['average']}/10")


if RUN_BATTLE:
    calls_before = tracker.summary()["TOTAL"]["llm_calls"]
    saved = (MAX_ROUNDS, TONE)
    MAX_ROUNDS, TONE = 1, "Keep it warm, fun and kind."  # one round + neutral tone for the rival, to save calls
    try:
        print(f"🥊 Rival: {RIVAL_IDEA['idea']}\n")
        rival = studio.invoke({"idea": RIVAL_IDEA["idea"], "sector": RIVAL_IDEA["sector"], "round": 0})
    finally:
        MAX_ROUNDS, TONE = saved  # put our settings back

    judge = llm.with_structured_output(BattleVote, method="function_calling")
    prompts = [[{"role": "system", "content": investor_prompt(inv) + " Now pick the ONE startup you'd rather fund."},
                {"role": "user", "content": f"Startup A:\n{pitch_card(final)}\n\nStartup B:\n{pitch_card(rival)}"}]
               for inv in INVESTORS]
    votes = judge.batch(prompts, run_config("pitch_battle"))  # all three vote at once

    names = {"A": final["brand"].startup_name, "B": rival["brand"].startup_name}
    print("\n🥊 PITCH BATTLE")
    for inv, v in zip(INVESTORS, votes):
        print(f"   {inv['name']} votes {names[v.winner]}: {v.reason}")
    a_votes = sum(v.winner == "A" for v in votes)
    winner = names["A"] if a_votes >= 2 else names["B"]
    print(f"\n🏆 Winner: {winner} ({max(a_votes, 3 - a_votes)}-{min(a_votes, 3 - a_votes)})")
    print(f"   Battle cost: {tracker.summary()['TOTAL']['llm_calls'] - calls_before} extra LLM calls")

    launch_kit["pitch_battle"] = {
        "rival_idea": RIVAL_IDEA["idea"], "rival_name": names["B"],
        "votes": [{"investor": inv["name"], "voted_for": names[v.winner], "reason": v.reason}
                  for inv, v in zip(INVESTORS, votes)],
        "winner": winner,
    }
````
